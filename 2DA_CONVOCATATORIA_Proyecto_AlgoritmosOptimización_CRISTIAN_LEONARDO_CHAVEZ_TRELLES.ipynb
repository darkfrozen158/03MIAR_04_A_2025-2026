# ***¿Cuantas posibilidades hay sin tener en cuenta las restricciones?***

**RESPUESTA:**
El numero de posibilidades estan relacionado con la cantidad de elementos, en este caso la cantidad de tomas que son 30.

Por tanto la cantidad de posibilidades sin restricciones es: n!
El resultado para n = 30 es: 30! = 2,6525285981219105863630848e+32

Ahora utilizamos los numeros de BELL para tener en cuenta el número total de formas posibles de particionar un conjunto de
n elementos en subconjuntos no vacíos (sin restricciones).

image.png

Consideraciones :
*   Tomas: 30 tomas distintas (numeradas del 1 al 30).

*   Días: Las tomas se agrupan en días, donde cada día es un subconjunto no vacío de tomas.

*   Sin restricciones: No hay límite en el número de tomas por día ni en la compatibilidad de actores.

**IMPLEMENTACIÓN EN CODIGO PYTHON:**

```python
def numeros_bell(n):
    """Calcula el número de Bell B_n usando programación dinámica."""
    if n == 0:
        return 1
    # Inicializar una tabla para almacenar los números de Stirling de segunda clase
    S = [[0] * (n + 1) for _ in range(n + 1)]
    S[0][0] = 1  # Caso base

    for i in range(1, n + 1):
        for k in range(1, i + 1):
            S[i][k] = k * S[i - 1][k] + S[i - 1][k - 1]

    # El número de Bell es la suma de la fila S[n]
    return sum(S[n])
```

# **¿Cuantas posibilidades hay teniendo en cuenta todas las restricciones?**




**RESPUESTA:**

Recordemos que el problema como tal tiene las siguientes restricciones:

*  Los actores del doblaje deben coincidir en las
tomas en las que sus personajes aparecen juntos en las diferentes tomas.

*  Los actores de
doblaje cobran todos la misma cantidad por cada día que deben desplazarse hasta el estudio de
grabación independientemente del número de tomas que se graben

*  No es posible grabar más
de 6 tomas por día.

Ademas el número de posibilidades válidas es extremadamente grande pero difícil de calcular exactamente.

Es una combinación de 30 tomas que pueden ser escogidas de 6 tomas en 6 tomas (tomas por día),
Donde las posibles muestras sin orden de las 6 tomas distintas que se pueden extraer del totas de 30 tomas.

Por tanto la cantidad de posibilidades con restricciones se calcula con la fórmula: n! / (r!*(n-r)!)

Para la pregunta n = 30 y r = 6.  
Aplicando la formula: 30!/(6!*(30-6)!)
El resultado es: 593,775 posibilidades

Dado que el cálculo exacto es inviable, usaremos una aproximación basada en permutaciones y poda:

**IMPLEMENTACIÓN EN CODIGO PYTHON:**

**1.DESARROLLAMOS LA FUNCIÓN**

```python
tomas = []
actores_por_toma = []
for _, row in df.iterrows():
    tomas.append(row['TOMA'])
    actores_por_toma.append([actor for actor in row.index[1:] if row[actor] == 1])

# Construir matriz de compatibilidad
n = len(tomas)
compatible = [[False] * n for _ in range(n)]
for i in range(n):
    for j in range(n):
        compatible[i][j] = len(set(actores_por_toma[i]) & set(actores_por_toma[j])) > 0

# Función para verificar si un grupo de tomas es válido (compatible y ≤6 tomas)
def es_grupo_valido(grupo):
    for i, j in combinations(grupo, 2):
        if not compatible[i][j]:
            return False
    return len(grupo) <= 6

# Estimación del número de soluciones (aproximación muy simplificada)
# Usamos una permutación aleatoria y contamos grupos válidos (Monte Carlo)
def estimar_soluciones(num_muestras=1000):
    conteo = 0
    for _ in range(num_muestras):
        perm = random.sample(range(n), n)
        dias = []
        grupo_actual = []
        for toma in perm:
            grupo_actual.append(toma)
            if not es_grupo_valido(grupo_actual):
                dias.append(grupo_actual[:-1])
                grupo_actual = [toma]
        if grupo_actual:
            dias.append(grupo_actual)
        if all(len(d) <= 6 for d in dias):
            conteo += 1
    return (conteo / num_muestras) * factorial(n)  # Ajuste por todas las permutaciones
```

**APLICAMOS LA FUNCIÓN:**

```python

# Ejemplo de estimación (muy lento para n=30)
print("Estimación del número de soluciones válidas:", estimar_soluciones(num_muestras=10))
```

# **Calcula la complejidad del algoritmo por fuerza bruta**

**RESPUESTA:**

Debido a que tenemos una gran cantidad de For anidados y estructuras se calcula que la Big(O) de este algoritmo por fuerza bruta es Polinomial O(n^6)



Se calcula de la siguiente manera:

1.   Para la parte 1 (asignacion de variables)
  O(1) + O(1) + O(1) = O(1)
1.   para la parte 2 y parte 3 misma estructura (2 for e if)
  O(n) * O(n) + O(n) = O(n^2) + O(n)
1.   para la parte 4  (6 bucles for anidados)
  O(n) * O(n) * O(n) * O(n) * O(n) * O(n) = O(n^6)
1.   para la parte 5  (2 bucles for anidados y 1 if)
  O(n) * O(n) + O(n) = O(n^2) + O(n)
1.   para la parte 6 (4 bucles for anidados)
  O(n) * O(n) * O(n) * O(n) = O(n^4)
1.   para la parte 7 (for e if)
  O(n) + O(n) = O(n^2)
  

**NOTA: En caso se aumentan mas restricciones en el caso de uso, los bucles For tendrian que aumentar proporcionalemente segun las restricciones dadas.**
